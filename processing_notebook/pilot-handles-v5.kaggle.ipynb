````python
import json
def norm_eng(s):
    return "".join(ch for ch in s.lower() if ch.isalnum())
def score_rows(rows, preds):
    agg = {"n": 0, "commenter_exact": 0, "english_match": 0, "exp_handles": 0, "leak": 0, "fp": 0, "clean_empty": 0, "empty_total": 0}
    details = []
    for r, p in zip(rows, preds):
        s = r["silver"]
        exp_c, exp_e, exp_m = s["commenter_tamil"], s["commenter_english"], s["mentioned"]
        got_c = p.get("commenter_tamil", [])
        got_e = p.get("commenter_english", [])
        has_exp = bool(exp_c)
        exact = got_c == exp_c
        eng = [norm_eng(x) for x in got_e] == [norm_eng(x) for x in exp_e]
        leak = bool(set(got_c) & set(exp_m))
        agg["n"] += 1
        agg["commenter_exact"] += exact
        if has_exp:
            agg["exp_handles"] += 1
            agg["english_match"] += eng
        else:
            agg["empty_total"] += 1
            agg["clean_empty"] += not got_c
            agg["fp"] += bool(got_c) and not leak
        agg["leak"] += leak
        details.append({"set": r["set"], "video_id": r["video_id"], "center": r["center_start"], "exact": exact, "eng": eng, "leak": leak, "pred": p, "note": s["note"]})
    return agg, details
````

````python
import json, time
OUTPUT_PATH = WORKING_DIR + f"/pilot_handles_v5_shard{SHARD_ID}.jsonl"
def llm_extract(text):
    prompt = SYS.replace("{TEXT}", text)
    msgs = [{"role": "user", "content": [{"type": "text", "text": prompt}]}]
    try:
        _dev = model.language_model.embed_tokens.weight.device
    except Exception:
        _dev = "cuda:0"
    inputs = tokenizer.apply_chat_template(msgs, add_generation_prompt=True, tokenize=True, return_dict=True, return_tensors="pt").to(_dev)
    out = model.generate(**inputs, max_new_tokens=128, do_sample=False)
    gen = tokenizer.decode(out[0][inputs.input_ids.shape[1]:], skip_special_tokens=True)
    cleaned = gen.replace("json", "").replace("```", "").strip()
    try:
        p = json.loads(cleaned)
        return {"commenter_tamil": p.get("commenter_tamil", []), "commenter_english": p.get("commenter_english", []), "mentioned": p.get("mentioned", [])}, False
    except Exception:
        return {"commenter_tamil": [], "commenter_english": [], "mentioned": []}, True
t0 = time.perf_counter()
preds, fails = [], 0
out = open(OUTPUT_PATH, "w", encoding="utf-8")
for ri, r in enumerate(ROWS):
    wtext = " / ".join(w["text"] for w in r["window"])
    p, f = llm_extract(wtext)
    fails += f
    preds.append(p)
    json.dump({"set": r["set"], "video_id": r["video_id"], "center_start": r["center_start"], "pred": p, "silver": r["silver"]}, out, ensure_ascii=False)
    out.write("\n")
    print(f"{ri + 1}/{len(ROWS)} fail={f} pred={json.dumps(p, ensure_ascii=False)}", flush=True)
out.close()
agg, details = score_rows(ROWS, preds)
reg_rows = [r for r, d in zip(ROWS, details) if r["set"] == "reg"]
cue_rows = [r for r, d in zip(ROWS, details) if r["set"] == "cue3"]
reg_agg, _ = score_rows(reg_rows, [preds[i] for i, r in enumerate(ROWS) if r["set"] == "reg"])
cue_agg, _ = score_rows(cue_rows, [preds[i] for i, r in enumerate(ROWS) if r["set"] == "cue3"])
agg["fails"] = fails
agg["secs"] = round(time.perf_counter() - t0, 1)
print("ALL:", json.dumps(agg, ensure_ascii=False), flush=True)
print("REG:", json.dumps(reg_agg, ensure_ascii=False), flush=True)
print("CUE3:", json.dumps(cue_agg, ensure_ascii=False), flush=True)
with open(WORKING_DIR + "/pilot_handles_v5_metrics.json", "w", encoding="utf-8") as mf:
    json.dump({"agg": agg, "reg": reg_agg, "cue3": cue_agg, "details": details}, mf, ensure_ascii=False, indent=1)
print(f"saved {OUTPUT_PATH}", flush=True)
````
